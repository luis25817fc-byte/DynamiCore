```python
"""
DynamiCore Enterprise
State Object v1.0

Todo análisis realizado por DynamiCore produce un único
DynamiCoreState.

Este objeto es el contrato oficial entre:

Engine
API
Dashboard
Research Lab
Benchmarks
AI
Persistence

No debe romper compatibilidad hacia adelante.
"""

from __future__ import annotations

from dataclasses import dataclass, field
from datetime import datetime
from typing import Any, Dict, List, Optional
import uuid


# ==========================================================
# ENGINE INFORMATION
# ==========================================================

ENGINE_NAME = "DynamiCore Enterprise"

ENGINE_VERSION = "1.0.0-alpha"


# ==========================================================
# METRIC CONTAINERS
# ==========================================================

@dataclass
class EntropyState:

    shannon: float = 0.0
    normalized: float = 0.0
    temporal: float = 0.0
    conditional: float = 0.0
    incremental: float = 0.0
    cross_entropy: float = 0.0


@dataclass
class MetricsState:

    R: float = 0.0
    delta_R: float = 0.0
    psi: float = 0.0
    divergence: float = 0.0


@dataclass
class BasinState:

    count: int = 0

    largest: int = 0

    entropy: float = 0.0

    resilience: float = 0.0

    basins: Dict[str, Any] = field(default_factory=dict)


@dataclass
class CycleState:

    count: int = 0

    longest: int = 0

    average_length: float = 0.0

    cycles: List[Any] = field(default_factory=list)


@dataclass
class GraphState:

    nodes: int = 0

    edges: int = 0

    density: float = 0.0

    connected_components: int = 0

    graph: Dict[str, Any] = field(default_factory=dict)


@dataclass
class MemoryState:

    history_length: int = 0

    previous_state: Optional[str] = None

    snapshots: int = 0


@dataclass
class DiagnosticsState:

    execution_ms: float = 0.0

    memory_mb: float = 0.0

    warnings: List[str] = field(default_factory=list)

    errors: List[str] = field(default_factory=list)


# ==========================================================
# ROOT OBJECT
# ==========================================================

@dataclass
class DynamiCoreState:

    id: str = field(default_factory=lambda: str(uuid.uuid4()))

    timestamp: str = field(
        default_factory=lambda: datetime.utcnow().isoformat()
    )

    engine: str = ENGINE_NAME

    version: str = ENGINE_VERSION

    dataset: str = ""

    source: str = ""

    input_size: int = 0

    metadata: Dict[str, Any] = field(default_factory=dict)

    entropy: EntropyState = field(default_factory=EntropyState)

    metrics: MetricsState = field(default_factory=MetricsState)

    basins: BasinState = field(default_factory=BasinState)

    cycles: CycleState = field(default_factory=CycleState)

    graph: GraphState = field(default_factory=GraphState)

    memory: MemoryState = field(default_factory=MemoryState)

    diagnostics: DiagnosticsState = field(
        default_factory=DiagnosticsState
    )

    def to_dict(self):

        return {

            "id": self.id,

            "timestamp": self.timestamp,

            "engine": self.engine,

            "version": self.version,

            "dataset": self.dataset,

            "source": self.source,

            "input_size": self.input_size,

            "metadata": self.metadata,

            "entropy": vars(self.entropy),

            "metrics": vars(self.metrics),

            "basins": vars(self.basins),

            "cycles": vars(self.cycles),

            "graph": vars(self.graph),

            "memory": vars(self.memory),

            "diagnostics": vars(self.diagnostics),
        }

    def summary(self):

        return {

            "Entropy": self.entropy.shannon,

            "R(k)": self.metrics.R,

            "ΔR(k)": self.metrics.delta_R,

            "Ψ(k)": self.metrics.psi,

            "D(k)": self.metrics.divergence,

            "Nodes": self.graph.nodes,

            "Cycles": self.cycles.count,

            "Basins": self.basins.count,

        }
```

```python
from pathlib import Path
import shutil
from datetime import datetime

ROOT = Path("/content/DynamiCore")
CORE = ROOT / "app" / "core"

assert CORE.exists(), "No existe app/core"

backup = ROOT / "_enterprise_backups" / datetime.now().strftime("%Y%m%d_%H%M%S")

backup.mkdir(parents=True, exist_ok=True)

files = [
    "engine.py",
    "entropy.py",
    "metrics.py",
    "graph.py",
    "cycles.py",
    "basins.py",
    "memory.py",
    "analyzer.py"
]

print("="*80)
print("CREANDO BACKUP DEL NÚCLEO")
print("="*80)

for f in files:
    src = CORE / f
    if src.exists():
        shutil.copy2(src, backup / f)
        print("✓", f)

print("\nBackup creado en:")
print(backup)
```
